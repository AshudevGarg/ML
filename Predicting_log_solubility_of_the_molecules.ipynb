#**Train ML Model On dataset 'delaney_solubility_with_descriptors'**

#**Load Data**

```python
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/dataprofessor/data/refs/heads/master/delaney_solubility_with_descriptors.csv')
```

```python
df
```

```python
df.head()
```

```python
y=df['logS']
```

#**Data Preparation**

##**Data Separation as X and Y**

```python
x=df.drop('logS',axis=1)
x
```

```python
y
```

##**Data Splitting**

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=100)
```

```python
x_train
```

```python
x_test
```

#**Model Building**


## **Linear Regression**


```python
from sklearn.linear_model import LinearRegression
```

###**Training the model**


```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

###**Applying the model to make a prediction**


```python
y_lr_train_pred = lr.predict(x_train)
y_lr_test_pred = lr.predict(x_test)
```

```python
y_lr_train_pred
```

```python
y_lr_test_pred
```

```python
y_train
```

###**Evaluate model performance**


```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
lr_train_mse=mean_squared_error(y_train,y_lr_train_pred)
lr_train_r2_score=r2_score(y_train,y_lr_train_pred)

lr_test_mse=mean_squared_error(y_test,y_lr_test_pred)
lr_test_r2_score=r2_score(y_test,y_lr_test_pred)
```

```python
lr_train_mse
```

```python
lr_test_mse
```

```python
lr_train_r2_score
```

```python
lr_test_r2_score
```

```python
lr_results=pd.DataFrame(['Linear Regression',lr_train_mse,lr_train_r2_score,lr_test_mse,lr_test_r2_score]).transpose()
lr_results.columns=['Method','Training MSE','Training R2','Test MSE','Test R2']
```

```python

lr_results
```

## **Random Forest**

```python
from sklearn.ensemble import RandomForestRegressor
```

### **Training the model**

---



```python
rf=RandomForestRegressor(max_depth=2,random_state=100)
rf.fit(x_train,y_train)
```

### **Applying the model to make a prediction**

```python
y_rf_train_pred=rf.predict(x_train)
y_rf_test_pred=rf.predict(x_test)
```

###**Evaluate model performance**

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
rf_train_mse=mean_squared_error(y_train,y_rf_train_pred)
rf_train_r2_score=r2_score(y_train,y_rf_train_pred)

rf_test_mse=mean_squared_error(y_test,y_rf_test_pred)
rf_test_r2_score=r2_score(y_test,y_rf_test_pred)
```

```python
rf_results=pd.DataFrame(['Random Forest',rf_train_mse,rf_train_r2_score,rf_test_mse,rf_test_r2_score]).transpose()
rf_results.columns=['Method','Training MSE','Training R2','Test MSE','Test R2']
```

```python
rf_results
```

## **Model Comparison**

```python
df_models=pd.concat([lr_results,rf_results],axis=0).reset_index(drop=True)
df_models
```

#**Data Visualization of Prediction Results**


```python
import matplotlib.pyplot as plt
import numpy as np
plt.scatter(x=y_train,y=y_lr_train_pred,alpha=0.3)
z=np.polyfit(y_train,y_lr_train_pred,1)
p=np.poly1d(z)
plt.plot(y_train,p(y_train),"r")
plt.xlabel('Experimental logS')
plt.ylabel('Predicted logS')
plt.title('Linear Regression Model')
```
